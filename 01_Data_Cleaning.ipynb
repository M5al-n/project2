## Introduction 
__Introduction to the topic__ 

  There are several variations among properties within the real estate market of Saudi Arabia that involve price, size, location, and features. The knowledge of these variations can provide insight into the patterns within property listing.

This study applies Exploratory Data Analysis (EDA) on the AQAR dataset that consists of 3,718 listings and 24 columns from four cities in Saudi Arabia: Riyadh, Jeddah, Dammam, and Khobar.

This analysis involves the analysis of the variation in prices between cities and districts as well as the relationship between property characteristics and prices. These property characteristics include size, bedroom, age, swimming pool, and frontyard.

***Data Handling:***
Cleaning, transforming, and combining data

```python
df.isnull().sum()
```

```python
df[["size", "bedrooms", "bathrooms", "price", "property_age"]].describe()
```

```python
size_q1 = df["size"].quantile(0.25)
size_q3 = df["size"].quantile(0.75)
size_iqr = size_q3 - size_q1

size_lower = size_q1 - 1.5 * size_iqr
size_upper = size_q3 + 1.5 * size_iqr

size_outliers = df[(df["size"] < size_lower) | (df["size"] > size_upper)]

print("Lower bound:", size_lower)
print("Upper bound:", size_upper)
print("Size outliers:", len(size_outliers))
print("Smallest flagged size:", size_outliers["size"].min())
print("Largest flagged size:", size_outliers["size"].max())
```

```python
df_clean = df.copy()
```

```python
df_clean.loc[[997, 1126], ["city", "size", "price", "details"]]
```

```python
df_clean = df_clean.drop(index=[997, 1126, 1971]).copy()
df_clean.shape
```

```python
df_clean["city"] = df_clean["city"].str.strip()
df_clean["district"] = df_clean["district"].str.strip()
df_clean["city"].value_counts()
```

```python
monthly = df_clean["details"].str.contains("شهري", na=False)
yearly = df_clean["details"].str.contains("سنوي", na=False)
daily = df_clean["details"].str.contains("يومي", na=False)

print("Monthly:", monthly.sum())
print("Annual:", yearly.sum())
print("Daily:", daily.sum())
```

```python
period_count = monthly.astype(int) + yearly.astype(int) + daily.astype(int)

print(period_count.value_counts().sort_index())
print("Total rows:", len(df_clean))
```

```python
df_clean["rental_type"] = "Unknown"
df_clean.loc[monthly & (period_count == 1), "rental_type"] = "Monthly"
df_clean.loc[yearly & (period_count == 1), "rental_type"] = "Annual"
df_clean.loc[daily & (period_count == 1), "rental_type"] = "Daily"
df_clean.loc[period_count > 1, "rental_type"] = "Multiple"
df_clean["rental_type"].value_counts()
```

```python
df_clean.groupby("rental_type")["price"].agg(["count", "min", "median", "max"])
```

```python
df_analysis = df_clean.copy()

df_analysis["annual_price"] = df_analysis["price"]

daily_mask = df_analysis["price"] <= 1500
monthly_mask = (df_analysis["price"] > 1500) & (df_analysis["price"] < 12000)

df_analysis.loc[daily_mask, "annual_price"] *= 365
df_analysis.loc[monthly_mask, "annual_price"] *= 12
```

```python
df_analysis[["city", "size", "price", "annual_price"]].sort_values("price").head(10)
```

```python
df_analysis["annual_price"].describe()
```

```python
price_q1 = df_analysis["annual_price"].quantile(0.25)
price_q3 = df_analysis["annual_price"].quantile(0.75)
price_iqr = price_q3 - price_q1

price_lower = price_q1 - 1.5 * price_iqr
price_upper = price_q3 + 1.5 * price_iqr

price_outliers = df_analysis[
    (df_analysis["annual_price"] < price_lower) |
    (df_analysis["annual_price"] > price_upper)
]

print("Lower bound:", price_lower)
print("Upper bound:", price_upper)
print("Price outliers:", len(price_outliers))

price_outliers[["city", "size", "annual_price"]].sort_values(
    "annual_price", ascending=False
).head(10)
```
